```python
import numpy as np 
```

```python
Psi = np.array([0,1,1,0])/np.sqrt(2)
```

```python
A0 = np.array([ [ 1, 0 ], [0, 1] ])
A1 = np.array([ [ 1, 1 ], [1, -1] ])/np.sqrt(2)
```

```python
B0 = np.array([ [ 1-np.sqrt(2), 1+np.sqrt(2) ], [1, 1] ])
B0 = B0 / np.linalg.norm(B0,axis=0)

B1 = np.array([ [ -1-np.sqrt(2), -1+np.sqrt(2) ], [1, 1] ])
B1 = B1 / np.linalg.norm(B1,axis=0)
B0, B1 
```

```python
p_bell = np.zeros((4,2,2))
```

```python
p_bell[:,0,0] = np.abs( np.kron(A0,B0)@Psi  )**2
p_bell[:,0,1] = np.abs( np.kron(A0,B1)@Psi  )**2
p_bell[:,1,0] = np.abs( np.kron(A1,B0)@Psi  )**2
p_bell[:,1,1] = np.abs( np.kron(A1,B1)@Psi  )**2
```
