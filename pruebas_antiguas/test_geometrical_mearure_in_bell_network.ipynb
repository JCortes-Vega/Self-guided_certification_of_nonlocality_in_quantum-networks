```python
import numpy as np
import matplotlib.pyplot as plt
from codes.Quantum import *
from codes.CSPSA import *
```

```python
psi = np.array([1, 0 , 0 , 1])/np.sqrt(2)
Psi = triple_kron( psi, psi, psi )
```

```python
measure = RandomState( 4 )
gains = [3, 0, 0.1, 0.602, 0.101]
max_iter = 100
```

```python
cspsa = CSPSA(gains=gains)
```

```python
def obj_fun( m ):
    mkron = MKron( m )
    return Infidelity( mkron, Psi )
```

```python
def postprocessing( x ):
    return x/np.linalg.norm(x)
```

```python
obj_fun( measure )
```

```python
results = cspsa.minimize( obj_fun, measure, max_iter, postprocessing=postprocessing )
```

```python
infidelities = results.fun 
```

```python
plt.plot( infidelities )
```
