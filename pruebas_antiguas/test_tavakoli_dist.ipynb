```python
import numpy as np 
import matplotlib.pyplot as plt
from codes.Quantum import *
from codes.CSPSA import *
```

```python
psi = np.sqrt(1/2)*np.array([0, 1, 0, 0]) + np.sqrt(1/2)*np.array([0, 0, 1, 0])
Psi = triple_kron( psi, psi, psi )
Psi = Psi.reshape(6*[2]).transpose([5,0,1,2,3,4]).flatten() 
Rho = np.outer( Psi, Psi.conj() )
```

```python
c = np.sqrt( (2-np.sqrt(2))/4 )
s = np.sqrt(1-c**2)
ket0 = np.array([1,0,0,0])
ket10 = np.array([0,c,s,0])
ket11 = np.array([0,s,-c,0])
ket2 = np.array([0,0,0,1])

m = np.array( [ ket0, ket10, ket11, ket2 ]).T
M = triple_kron( m, m , m ) 
```

```python
probs = np.abs( M.T.conj() @ Psi )**2
probs = probs.reshape(4,4,4)
probs.shape 
```
