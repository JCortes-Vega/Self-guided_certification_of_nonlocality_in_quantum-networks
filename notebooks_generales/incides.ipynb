Primero definimos las mediciones. Sean $\{A_0,A_1 \}$ los observables de Alice y $\{B_0, B_1\}$ los de Bob. Sea $\{ |a_{ij}\rangle\}$ los autoestados de los observables $\{A_l\}$ y $\{ |b_{ij}\rangle\}$ los de $\{B_m\}$. Así 
$$ M[0,l,k,:]= |a_{lk}\rangle \quad \text{y}  \quad M[1,m,k',:]= |b_{mk'}\rangle.$$ 

```python
M = np.zeros( (2,2,2,2) )
# indices Alice(0) or Bob(1), Obs 0 o 1, autovec 0 o 1, largo del vector
```

```python
M[0,1,:,:] = np.array([  [1, 0], [0, 1]  ]) #Z
M[0,0,:,:] = np.array([  [1, 1], [1, -1]  ])/np.sqrt(2) #X
```

```python
M[1,0,0,:] = np.array([1-np.sqrt(2), 1])/np.sqrt( 2*(2-np.sqrt(2)) )  #X+Z
M[1,0,1,:] = np.array([1+np.sqrt(2), 1])/np.sqrt( 2*(2+np.sqrt(2)) )

M[1,1,0,:] = np.array([-1-np.sqrt(2), 1])/np.sqrt( 2*(2+np.sqrt(2)) )
M[1,1,1,:] = np.array([-1+np.sqrt(2), 1])/np.sqrt( 2*(2-np.sqrt(2)) ) #X-Z
```

```python
M
```

Ahora calcularemos la distribución de probabilidad
$$ P(klk'mij) = \frac{1}{4} P_{Bell}(kk'|lm)\delta_{li}\delta_{mj},$$
con
$$ P_{Bell}(kk'|lm) = Tr [ \rho |a_{lk}\rangle\langle a_{lk}|\otimes |b_{mk'}\rangle\langle b_{mk'}| ]. $$

```python
probs_triangle = np.zeros( (2,2,2,2,2,2) )
Psi_ab = np.array([ 0, 1, -1, 0 ])/np.sqrt(2)

for k in range(2):
    for kk in range(2):
        for l in range(2):
            for m in range(2):
                measure = np.kron( M[0,l,k,:], M[1,m,kk,:] )  
                probs_triangle[ l, k, m, kk, l, m ] = 0.25 * np.abs( np.vdot( measure, Psi_ab ) )**2 
```

```python
probs_triangle
```

Podemos ver que las distribución de probabilidad $P$ coincide con la distribución de Fritz.

```python
# alpha alpha beta beta gamma gamma
# a     b     b    c    c     a 
# 0     1     2    3    4     5
# -----> 
# gamma alpha beta alpha beta gamma 
# a     a     b     b    c     c
# 5     0     2     1    3     4   
```

```python
Psi = triple_kron( Psi_ab, Psi_ab, Psi_ab )
Psi = Psi.reshape(6*[2]).transpose([5,0,2,1,4,3]).flatten() 
Rho = np.outer( Psi, Psi.conj() )
```

```python
Comp_basis = np.eye(2) 

vec_alice = np.zeros( (2,2,4,4) )
for l in range(2):
    for k in range(2):
        operator  = np.kron( Comp_basis[l], M[0,l,k,:] )
        vec_alice[l,k,:,:] = np.outer( operator, operator.conj() )
vec_alice = vec_alice.reshape(4,4,4)
```

```python
vec_bob = np.zeros( (2,2,4,4) )
for m in range(2):
    for kk in range(2):
        operator = np.kron( Comp_basis[m], M[1,m,kk,:] )
        vec_bob[m,kk,:,:] = np.outer( operator, operator.conj() )
vec_bob = vec_bob.reshape(4,4,4)
```

```python
Comp_basis = np.array([ 0,1,1,0 ]).reshape(2,2)

vec_charlie = np.zeros( (2,2,4,4) )
for i in range(2):
    for j in range(2):
        operator = np.kron( Comp_basis[i], Comp_basis[j] )
        vec_charlie[i,j,:,:] = np.outer( operator, operator.conj() ) 
vec_charlie = vec_charlie.reshape(4,4,4)
```

```python
probs_direct = InnerProductMatrices( Rho, 
                                    [ vec_alice, 
                                        vec_bob, 
                                        vec_charlie] 
                                    ).real.reshape(2,2,2,2,2,2)
```

Wagon-Wheel Inequality 

```python
probs = probs_direct
probs_albl = np.sum( probs, axis=(1,3,4,5) )
probs_alblclcr = np.sum( probs, axis=(1,3) )
probs_clcr = np.sum( probs, axis=(0,1,2,3) )
```

```python
inequality = (-probs_albl[1,0] 
                +probs_alblclcr[1,0,1,0] 
                -probs_albl[0,1] * probs_clcr[1,0] 
                -probs_clcr[0,0] * probs_clcr[1,1] 
                -probs_clcr[0,1] * probs[1,0,0,1,1,0]
                -probs_clcr[0,1] * probs[1,1,0,0,1,0]
                +probs_clcr[0,0] * probs[1,0,1,1,1,1]
                +probs_clcr[0,0] * probs[1,1,1,0,1,1]
                +probs_clcr[1,0] * probs[0,0,1,0,0,1]
                +probs_clcr[1,0] * probs[0,1,1,1,0,1]
                +probs_clcr[1,1] * probs[0,0,0,0,0,0]
                +probs_clcr[1,1] * probs[0,1,0,1,0,0]
                )
```
